# FedGuard - Kaggle helper machine

Runs one share of the FedGuard experiment queue on Kaggle's free GPU (guide: `docs/FedGuard_Faster_Training_Guide.pdf` in the repository).

Before running: **Settings -> Accelerator = GPU T4 x2**, **Internet = On**, and add your private dataset that contains `nf_unsw.npz` (+ `nf_unsw_manifest.json`) with **Add Input**.

First time: set `MACHINE = "smoke"` and run all cells interactively (about 15 minutes). Then set the real share and use **Save Version -> Save & Run All (Commit)**.

```python
# Which share of the queue this notebook runs (ask the main-laptop owner).
MACHINE = "smoke"     # e.g. "2" (machine 2), "2,3" (machines 2 and 3, one per T4 GPU), or "smoke"
MACHINES = "3"        # total number of machines sharing the queue
HOURS = "10.5"        # no new run starts after this many hours (Kaggle stops sessions at 12 h)
COMMIT = "main"       # the exact commit the main laptop uses (git log -1 --format=%h)
ONLY = "all"          # "dp" when a CPU-only laptop takes the runs without DP (guide 4.7), else "all"
REVERSE = "0"         # "1" = take runs from the END of the queue (main laptop works from the front)
```
